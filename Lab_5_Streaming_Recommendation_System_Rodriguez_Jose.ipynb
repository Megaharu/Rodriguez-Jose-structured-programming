```python
# Lab 5. Streaming Recommendation System
# Structured Programming Unit 2

def show_menu():
    """Muestra el menú principal al usuario."""
    print("\n=== NETFLIX ===")
    print("Select a genre:")
    print("1. Technology")
    print("2. Action")
    print("3. Science Fiction")

def get_genre():
    """Solicita la opción del usuario y la retorna."""
    option = input("Option: ")
    return option

def recommend_content(genre):
    """Recibe la opción del género y retorna una lista de recomendaciones."""
    if genre == '1':
        return ["The IT Crowd", "Silicon Valley", "Mr. Robot"]
    elif genre == '2':
        return ["Deadly Code", "Operation Delta", "John Wick"]
    elif genre == '3':
        return ["Interstellar", "The Matrix", "Dune"]
    else:
        return []

def rate_content():
    """Reto adicional: Permite calificar el contenido."""
    while True:
        try:
            score = int(input("Rate this content (1 to 5): "))
            if 1 <= score <= 5:
                print(f"Thank you! You rated the content with a {score}.")
                break
            else:
                print("Please enter a valid number between 1 and 5.")
        except ValueError:
            print("Invalid input. Please enter a whole number.")

def main():
    """Controla el flujo principal del programa."""
    while True:
        show_menu()
        user_option = get_genre()

        recommendations = recommend_content(user_option)

        if recommendations:
            print("\nRecommendations:")
            for item in recommendations:
                print(f"- {item}")

            print("")
            rate_content()
        else:
            print("\nInvalid option. Please select 1, 2, or 3.")

        again = input("\nWould you like to perform another search? (Y/N): ").strip().upper()
        if again != 'Y':
            print("Exiting program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
